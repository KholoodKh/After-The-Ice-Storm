```python
from mealpy.bio_based import SMA
import numpy as np
from greyhaven_problem import GreyhavenProblem
```

```python
problem = GreyhavenProblem("Evening pressure")

stations = [
    [1.5, 2.0, 1.2, 4, 1, 1],
    [3.0, 4.0, 1.5, 5, 2, 1],
    [6.0, 6.0, 1.0, 3, 1, 1],
    [8.0, 8.0, 1.8, 6, 3, 1]
]

problem.plot_deployment(stations)
```

```python
def fitness_function(solution):
    return np.sum(solution**2)

problem = {
    "fit_func": fitness_function,
    "lb": [-100, ] * 30,
    "ub": [100, ] * 30,
    "minmax": "min",
    "log_to": None,
    "save_population": False,
}

## Run the algorithm
model = SMA.BaseSMA(epoch=100, pop_size=50, pr=0.03)
best_position, best_fitness = model.solve(problem)
print(f"Best solution: {best_position}, Best fitness: {best_fitness}")
```

```python
print(best_position)
print(best_fitness)
print(model.get_parameters())
print(model.get_name())
print(model.get_attributes()["solution"])
print(model.problem.get_name())
print(model.problem.n_dims)
```

```python
import pkgutil, inspect, importlib
import mealpy
from mealpy.optimizer import Optimizer

found = {}
for m in pkgutil.walk_packages(mealpy.__path__, "mealpy."):
    if ".utils" in m.name:
        continue
    try:
        mod = importlib.import_module(m.name)
    except Exception:
        continue
    for name, cls in inspect.getmembers(mod, inspect.isclass):
        if issubclass(cls, Optimizer) and cls is not Optimizer and cls.__module__ == mod.__name__:
            found[name] = m.name.split(".")[1]   # category, e.g. swarm_based

print(len(found))
print([n for n in found])
```
